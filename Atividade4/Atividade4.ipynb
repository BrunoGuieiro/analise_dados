```python
import pandas as pd
```

```python
path = r"H:\Python 3Etapa\analise_dados\Atividade4\CD2022_Populacao_2010_Compatibilizada_20231222.xlsx"
```

```python
tabela = pd.read_excel(path, header=2)
```

```python
tabela
```

```python
tabela_2 = tabela.iloc[0:5570, 1:]
```

```python
tabela_2
```

```python
col_2010 = "População 2010 (Alterações de Limites até 2022)1"
col_2022 = "População Censo 2022"
```

```python
df_estado = (
    tabela_2.groupby("UF")[[col_2010, col_2022]]
    .sum()
    .reset_index()
)
df_estado["Diferença"] = df_estado[col_2022] - df_estado[col_2010]
df_estado = df_estado.sort_values("Diferença", ascending=False)
df_estado
```

```python
df_estado.to_csv(r"H:\Python 3Etapa\analise_dados\Atividade4\populacao_estado.csv", sep=";")
```

```python
df_municipio = tabela_2[["NOME DO MUNICÍPIO", col_2010, col_2022]].copy()
df_municipio["Diferença"] = df_municipio[col_2022] - df_municipio[col_2010]
df_municipio = df_municipio.sort_values("Diferença", ascending=False)
df_municipio
```

```python
df_municipio.to_csv(r"H:\Python 3Etapa\analise_dados\Atividade4\populacao_municipio.csv", sep=";")
```
